## 🛠️ Preparación de datos e ingeniería de atributos

### Registros duplicados

El conjunto de datos contiene 623 registros. Se identificaron 263 filas pertenecientes a grupos de registros exactamente duplicados y 152 copias adicionales. Al conservar una sola ocurrencia de cada registro, el conjunto de datos estaría conformado por 471 registros únicos.

Debido a que los registros duplicados presentan coincidencia en todas las variables, incluida `Chance of Admit`, su permanencia podría generar fuga de información si observaciones idénticas quedan distribuidas entre los conjuntos de entrenamiento y prueba. Por esta razón, se recomienda conservar una única ocurrencia de cada registro exactamente duplicado antes de realizar la partición de los datos.

### Valores nulos

Se identificaron valores nulos en diferentes variables del conjunto de datos. Antes del modelado será necesario definir una estrategia de tratamiento acorde con el tipo y comportamiento de cada atributo.

La imputación no deberá realizarse de manera automática sin considerar las características de cada variable. Para las variables numéricas podrán evaluarse medidas de tendencia central u otros métodos de imputación, mientras que para las variables ordinales y booleanas deberá conservarse su naturaleza durante el tratamiento.

### Valores atípicos

Durante el análisis exploratorio se identificaron algunas observaciones potencialmente atípicas mediante el criterio IQR y los boxplots. Sin embargo, estos valores se encuentran dentro de rangos posibles para las variables analizadas y no necesariamente corresponden a errores.

Por esta razón, no se recomienda eliminar automáticamente los valores atípicos. Su influencia deberá evaluarse posteriormente durante el desarrollo y evaluación de los modelos.

### Tipos de variables

Las variables `University Rating`, `SOP` y `LOR` presentan una naturaleza ordinal, por lo que su orden deberá conservarse durante la preparación de los datos.

La variable `Research` es booleana y podrá ser representada mediante una codificación binaria cuando el algoritmo utilizado requiera entradas numéricas.

### Escala de las variables

Las variables numéricas `GRE Score`, `TOEFL Score` y `CGPA` se encuentran en escalas diferentes. Dependiendo del algoritmo seleccionado, podrá ser necesario aplicar técnicas de normalización o estandarización.

### Relación entre variables predictoras

Se identificaron correlaciones altas entre `GRE Score`, `TOEFL Score` y `CGPA`, con valores entre 0.84 y 0.85. Esto indica que las variables comparten una cantidad importante de información.

Esta relación deberá considerarse durante la selección de características y la evaluación de los modelos, especialmente en algoritmos sensibles a relaciones fuertes entre variables predictoras.

## 🔄 Posibles transformaciones de los datos


### Eliminación de registros duplicados

Se recomienda eliminar las 152 copias adicionales de registros exactamente duplicados y conservar una única ocurrencia de cada registro. Esta transformación permitiría trabajar con 471 registros únicos y reducir el riesgo de fuga de información durante la división entre los conjuntos de entrenamiento y prueba.

### Tratamiento de valores nulos

Será necesario definir una estrategia de tratamiento para los valores faltantes. Dependiendo del tipo y distribución de cada variable, podrán evaluarse métodos de imputación apropiados, procurando no modificar innecesariamente las características originales de los datos.

### Codificación de variables

La variable `Research` podrá transformarse a una representación binaria, por ejemplo 0 y 1, cuando el algoritmo de machine learning seleccionado requiera variables numéricas.

Las variables `University Rating`, `SOP` y `LOR` deberán conservar su naturaleza ordinal durante cualquier proceso de codificación.

### Escalamiento de variables numéricas

Debido a que `GRE Score`, `TOEFL Score` y `CGPA` utilizan escalas diferentes, podrá evaluarse la aplicación de técnicas de estandarización o normalización para aquellos algoritmos que sean sensibles a la escala de las variables.

### Tratamiento de valores atípicos

Los valores potencialmente atípicos identificados no deberán eliminarse automáticamente. Antes de aplicar transformaciones sobre ellos será necesario evaluar su influencia en el desempeño de los modelos y comprobar que no correspondan a errores en los datos.

### Selección de características

Las correlaciones altas observadas entre `GRE Score`, `TOEFL Score` y `CGPA` sugieren que podría existir información redundante entre estas variables. Durante el modelado podrá evaluarse si es conveniente conservar todas las características o aplicar algún método de selección de variables.

Las transformaciones finalmente aplicadas deberán definirse de acuerdo con las características y requerimientos del modelo seleccionado.

## 🧠 Resolución manual del problema: modelo simple viable o heurística

Sí es posible plantear una solución manual para realizar una estimación inicial de la probabilidad de admisión de un aspirante.

A partir del análisis exploratorio se identificó que `CGPA`, `GRE Score` y `TOEFL Score` presentan las asociaciones más fuertes con `Chance of Admit`. También se observaron patrones relacionados con `University Rating`, `SOP`, `LOR` y `Research`.

Por lo tanto, una solución manual consistiría en evaluar primero el desempeño académico del aspirante mediante sus valores de CGPA, GRE y TOEFL y complementar esta evaluación con la valoración de la universidad, la declaración de propósito, la carta de recomendación y la experiencia en investigación.

Un aspirante que presente un desempeño alto de forma conjunta en las principales variables académicas y características favorables en las demás variables tendría una expectativa de admisión mayor. Por el contrario, un aspirante con valores bajos en varias de estas características tendría una expectativa de admisión menor. Los perfiles intermedios requerirían considerar conjuntamente todas las características disponibles.

Esta solución es viable como una aproximación simple e interpretable y puede utilizarse como referencia inicial para el problema. Sin embargo, presenta limitaciones, debido a que no permite determinar de manera objetiva el peso que debe tener cada característica, establecer con precisión la probabilidad de admisión ni representar adecuadamente todas las interacciones entre las variables.

## 📌 Conclusiones generales del análisis exploratorio

El análisis exploratorio permitió caracterizar el conjunto de datos de admisiones e identificar los principales aspectos que deberán considerarse antes de implementar un modelo de machine learning.

El conjunto de datos está compuesto por variables relacionadas con el desempeño académico y las características del aspirante, incluyendo `GRE Score`, `TOEFL Score`, `University Rating`, `SOP`, `LOR`, `CGPA` y `Research`, mientras que `Chance of Admit` corresponde a la variable objetivo del problema.

Durante el análisis univariable se identificaron valores nulos en diferentes atributos, registros exactamente duplicados y algunas observaciones potencialmente atípicas. En particular, se encontraron 152 copias adicionales de registros duplicados, por lo que al conservar una única ocurrencia se obtendrían 471 registros únicos. Este aspecto deberá ser tratado antes del modelado para reducir el riesgo de fuga de información entre los conjuntos de entrenamiento y prueba.

Los valores potencialmente atípicos identificados se encuentran dentro de rangos posibles para las variables analizadas, por lo que no se considera apropiado eliminarlos automáticamente. Su influencia deberá evaluarse posteriormente durante el desarrollo de los modelos.

El análisis bivariable mostró asociaciones positivas fuertes entre las principales variables académicas y `Chance of Admit`. `CGPA` presentó la mayor correlación con la variable objetivo (0.88), seguida de `GRE Score` (0.82) y `TOEFL Score` (0.81). Las variables `University Rating`, `SOP`, `LOR` y `Research` también mostraron patrones asociados con diferencias en la probabilidad de admisión.

El análisis multivariable permitió observar que la combinación de diferentes características proporciona una representación más completa del comportamiento de `Chance of Admit`. Los aspirantes con mejores resultados académicos y características favorables en otras variables tienden a concentrarse en niveles superiores de probabilidad de admisión.

También se identificaron correlaciones altas entre `GRE Score`, `TOEFL Score` y `CGPA`, con valores entre 0.84 y 0.85. Esto indica que estas variables comparten información y deberá considerarse durante las etapas posteriores de selección de características y evaluación de los modelos.

A partir del EDA se identificaron necesidades de preparación relacionadas con los registros duplicados, valores nulos, codificación de variables, conservación de la naturaleza ordinal de algunos atributos y posible escalamiento de las variables numéricas. Asimismo, se identificaron posibles transformaciones y datos adicionales que podrían evaluarse en etapas posteriores del proyecto.

Finalmente, los patrones encontrados permiten plantear una solución manual o heurística basada en la evaluación conjunta de las características académicas y complementarias del aspirante. Esta aproximación puede utilizarse como una referencia inicial o *baseline*, aunque presenta limitaciones para establecer objetivamente el peso de cada variable y estimar con precisión la probabilidad de admisión.
